# Preprocess

````python
%%writefile src/preprocess.py

import re
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory

stemmer = StemmerFactory().create_stemmer()

STOPWORDS = set(['yang', 'dan', 'di', 'dari', 'itu', 'adalah'])

def clean(text):
    """Melakukan case-folding dan normalisasi dasar."""
    text = text.lower()
    text = re.sub(r'[^\w\s]', ' ', text)
    text = re.sub(r'\s+', ' ', text).strip()
    return text

def tokenize(text):
    """Tokenisasi teks."""
    return text.split()

def remove_stopwords(tokens):
    """Menghapus stopword."""
    return [token for token in tokens if token not in STOPWORDS]

def stem(tokens):
    """Melakukan stemming pada token (menggunakan Sastrawi)."""
    return [stemmer.stem(token) for token in tokens]

def preprocess_document(text):
    text_cleaned = clean(text)
    tokens_raw = tokenize(text_cleaned)
    tokens_filtered = remove_stopwords(tokens_raw)
    tokens_stemmed = stem(tokens_filtered)

    return tokens_stemmed

if __name__ == '__main__':
    sample_text = "Fakultas Ilmu Komputer UDINUS akan mengadakan Kuliah Umum tentang Kecerdasan Buatan (Artificial Intelligence) pada tanggal 06 November 2025."

    print(f"Teks Asli: {sample_text}")

    processed_tokens = preprocess_document(sample_text)
    print(f"Token Hasil Preprocessing: {processed_tokens}")
````

````python
%%writefile src/boolean_ir.py
# src/boolean_ir.py: Implementasi Boolean Retrieval Model (Soal 03)

from collections import defaultdict
import numpy as np
# Catatan: Kita akan menggunakan list/dict sederhana dulu,
# NumPy dan sparse matrix (untuk Incidence Matrix) bisa ditambahkan
# pada implementasi yang lebih lanjut sesuai Soal 03[cite: 70].

class BooleanIR:
    """Kelas untuk mengelola Inverted Index dan menjalankan query Boolean."""
    def __init__(self, processed_docs):
        # processed_docs adalah dictionary {doc_id: [token1, token2, ...]}
        self.processed_docs = processed_docs
        self.vocabulary = set()
        self.inverted_index = self.build_inverted_index()

    def build_vocabulary(self):
        """Membangun himpunan semua term unik (vocabulary)."""
        for tokens in self.processed_docs.values():
            self.vocabulary.update(tokens)
        return self.vocabulary

    def build_inverted_index(self):
        """
        Membangun Inverted Index: {term: set(doc_id yang mengandung term)}.
        Sesuai Soal 03: dict(term → postings).
        """
        inverted_index = defaultdict(set)
        for doc_id, tokens in self.processed_docs.items():
            for term in set(tokens): # Gunakan set untuk menghindari duplikasi doc_id
                inverted_index[term].add(doc_id)

        # Simpan vocabulary setelah index dibuat
        self.vocabulary = set(inverted_index.keys())

        print(f"✅ Inverted Index berhasil dibangun. Total Vocabulary: {len(self.vocabulary)} term.")
        return dict(inverted_index)

    def query_boolean(self, query: str):
        """
        Parser query Boolean sederhana: mendukung AND/OR/NOT.
        (Implementasi parser yang kompleks  akan membutuhkan banyak kode).
        Di sini kita asumsikan format query sederhana: 'term1 AND term2', 'term1 OR NOT term2', dsb.
        """
        # CATATAN: Implementasi parser ini adalah tugas inti Soal 03
        # Kami sediakan kerangka dasar yang perlu Anda kembangkan!

        # Asumsikan kita hanya menangani satu operator untuk demo ini
        query = query.upper()

        if "AND" in query:
            op_term = "AND"
            parts = [p.strip() for p in query.split("AND")]
        elif "OR" in query:
            op_term = "OR"
            parts = [p.strip() for p in query.split("OR")]
        elif "NOT" in query:
            op_term = "NOT"
            # Asumsi NOT term1
            parts = ["", query.replace("NOT", "").strip()]
        else:
            op_term = "SINGLE"
            parts = [query.strip()]

        return self._execute_boolean_op(op_term, parts)

    def _execute_boolean_op(self, operator, terms):
        """Fungsi pembantu untuk menjalankan operasi Boolean pada posting list."""

        # Dapatkan semua Doc IDs yang ada (Universe)
        all_doc_ids = set(self.processed_docs.keys())

        # Kasus 1: Query Tunggal (tidak ada operator)
        if operator == "SINGLE":
            term = terms[0].lower()
            return self.inverted_index.get(term, set())

        # Kasus 2: NOT term
        if operator == "NOT":
            term = terms[1].lower()
            postings = self.inverted_index.get(term, set())
            return all_doc_ids.difference(postings)

        # Kasus 3: AND/OR
        term1 = terms[0].lower()
        term2 = terms[1].lower()

        set1 = self.inverted_index.get(term1, set())
        set2 = self.inverted_index.get(term2, set())

        if operator == "AND":
            # Interseksi
            return set1.intersection(set2)
        elif operator == "OR":
            # Union
            return set1.union(set2)

        return set() # Jika gagal

# Catatan: Fungsi build_incidence_matrix(docs) [cite: 70] belum diimplementasikan di sini,
# Anda bisa menambahkannya sebagai metode terpisah.
````

````python
%%writefile src/boolean_ir.py
# src/boolean_ir.py: Implementasi Boolean Retrieval Model (Soal 03 - V2)

from collections import defaultdict
import re

class BooleanIR:
    def __init__(self, processed_docs):
        self.processed_docs = processed_docs
        self.inverted_index = self._build_inverted_index()

    def _build_inverted_index(self):
        """Membangun Inverted Index: {term: set(doc_id yang mengandung term)}."""
        inverted_index = defaultdict(set)
        for doc_id, tokens in self.processed_docs.items():
            for term in set(tokens):
                inverted_index[term].add(doc_id)
        return dict(inverted_index)

    def _get_postings(self, term):
        """Mendapatkan posting list untuk term tertentu."""
        term = term.lower()
        # Jika term mengandung NOT
        if term.startswith("not "):
            base_term = term[4:].strip()
            # NOT adalah komplemen dari semua dokumen
            all_doc_ids = set(self.processed_docs.keys())
            postings = self.inverted_index.get(base_term, set())
            return all_doc_ids.difference(postings)

        # Term biasa
        return self.inverted_index.get(term, set())

    def query_boolean(self, query: str):
        """
        Parser Query Boolean yang disempurnakan.
        Prioritas (sederhana): AND > OR. Parentheses TIDAK wajib diimplementasikan
        di sini untuk menjaga kesederhanaan, fokus pada kombinasi AND/OR/NOT.
        """
        query = query.upper().strip()

        # 1. Handle NOT (dilakukan di _get_postings)
        # 2. Tokenisasi query (pisahkan berdasarkan operator)
        tokens = re.split(r' (AND|OR|NOT) ', query)

        # Proses pertama: AND
        current_set = None
        current_op = "AND" # Asumsi operator awal adalah AND untuk term pertama

        for i, token in enumerate(tokens):
            token = token.strip()
            if not token:
                continue

            if token == "AND":
                current_op = "AND"
            elif token == "OR":
                current_op = "OR"
            elif token == "NOT":
                 # Diabaikan karena _get_postings sudah menangani "NOT term"
                 continue
            else:
                # Ini adalah term (atau "NOT term")
                term = token.lower()

                # Dapatkan postings list
                postings = self._get_postings(term)

                if current_set is None:
                    current_set = postings
                else:
                    # Lakukan operasi
                    if current_op == "AND":
                        current_set = current_set.intersection(postings)
                    elif current_op == "OR":
                        current_set = current_set.union(postings)

        return current_set if current_set is not None else set()

    def get_explanation(self, query: str, result_docs: set):
        """
        Menghasilkan penjelasan sederhana untuk dokumen yang ditemukan.
        """
        explanation = []
        # Untuk query sederhana, kita hanya cek keberadaan term
        query_terms = [t.lower() for t in re.split(r' (AND|OR|NOT) ', query) if t.strip() not in ["AND", "OR", "NOT", ""]]

        for doc_id in result_docs:
            doc_tokens = set(self.processed_docs.get(doc_id, []))
            found_terms = [term for term in query_terms if term in doc_tokens]

            if found_terms:
                 explanation.append(f"Dokumen {doc_id.upper()} relevan karena mengandung term: {', '.join(found_terms)}.")
            else:
                 explanation.append(f"Dokumen {doc_id.upper()} relevan (Operasi Kompleks).") # Kasus NOT atau Query kompleks

        return explanation
````

````python
%%writefile src/vsm_ir.py
# src/vsm_ir.py: Implementasi Vector Space Model (VSM) dan Ranking (Soal 04)

import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
from scipy.sparse import csr_matrix
from collections import defaultdict

class VSM_IR:
    """Kelas untuk mengelola VSM, menghitung TF-IDF, dan Ranking."""

    def __init__(self, processed_docs):
        # processed_docs: dictionary {doc_id: [token1, token2, ...]}
        self.processed_docs = processed_docs
        self.doc_ids = list(processed_docs.keys())
        self.documents_text = [' '.join(tokens) for tokens in processed_docs.values()]

        # Inisialisasi TfidfVectorizer
        self.vectorizer = TfidfVectorizer()

        # 1. Bangun Matriks TF-IDF Dokumen (Soal 04, Langkah 1)
        self.doc_tfidf_matrix = self._build_tfidf_matrix()

        # Simpan Vocabulary (fitur names)
        self.vocabulary = self.vectorizer.get_feature_names_out()

        print(f"✅ VSM dibangun. Total Vocabulary: {len(self.vocabulary)} term.")
        print(f"   TF-IDF Matrix shape: {self.doc_tfidf_matrix.shape}")

    def _build_tfidf_matrix(self):
        """Menghitung TF, IDF, dan membentuk matriks TF-IDF."""
        # Fit dokumen ke vectorizer dan transform
        # Vectorizer secara otomatis menangani TF (Term Frequency)
        # dan IDF (Inverse Document Frequency)

        # NOTE: Output adalah sparse matrix (Soal 04, Langkah 1)
        tfidf_matrix = self.vectorizer.fit_transform(self.documents_text)
        return tfidf_matrix

    def transform_query(self, query: str) -> csr_matrix:
        """
        Representasi query sebagai vektor TF-IDF (Soal 04, Langkah 2).
        Query harus melalui preprocessing (seperti di Soal 02) sebelum dipanggil!
        """
        return self.vectorizer.transform([query])

    def rank_documents(self, query: str, k: int = 5):
        """
        Menghitung Cosine Similarity dan mengambil top-k (Soal 04, Langkah 3).
        """
        # 1. Ubah query menjadi vektor TF-IDF
        query_vector = self.transform_query(query)

        # 2. Hitung Cosine Similarity
        # Hasil: array 1D berisi skor kemiripan antara query dan setiap dokumen
        similarity_scores = cosine_similarity(query_vector, self.doc_tfidf_matrix).flatten()

        # 3. Dapatkan indeks dokumen yang memiliki skor tertinggi (ranking)
        # argpartition adalah cara cepat untuk mendapatkan k elemen tertinggi
        top_k_indices = np.argpartition(similarity_scores, -k)[-k:]

        # Dapatkan (skor, id dokumen) dari ranking teratas
        results = []
        for index in top_k_indices:
            doc_id = self.doc_ids[index]
            score = similarity_scores[index]
            results.append({'doc_id': doc_id, 'score': score, 'index': index})

        # Urutkan secara descending berdasarkan skor
        results.sort(key=lambda x: x['score'], reverse=True)

        return results
````

````python
import numpy as np
from typing import List, Set, Dict, Tuple

# --- Fungsi Evaluasi Ranking (Precision@k, AP, MAP@k) ---

def precision_at_k(retrieved_list: List[str], relevant_set: Set[str], k: int) -> float:
    """
    Menghitung Precision@k.
    Proporsi dokumen yang relevan di antara top-k yang diambil.
    """
    if k == 0:
        return 0.0

    # Ambil top-k dokumen yang diambil oleh sistem
    retrieved_k = retrieved_list[:k]

    # Hitung jumlah dokumen yang relevan di top-k
    relevant_in_k = len(set(retrieved_k).intersection(relevant_set))

    return relevant_in_k / k

def average_precision(retrieved_list: List[str], relevant_set: Set[str]) -> float:
    """
    Menghitung Average Precision (AP).
    Metrik ini hanya menghitung precision pada posisi di mana dokumen relevan ditemukan.
    """
    relevant_count = 0
    precision_sum = 0.0

    if not relevant_set:
        return 0.0

    for i, doc_id in enumerate(retrieved_list):
        k = i + 1 # Posisi saat ini
        if doc_id in relevant_set:
            relevant_count += 1
            # Hitung Precision@k pada titik ini dan jumlahkan
            precision_sum += precision_at_k(retrieved_list, relevant_set, k)

    # AP adalah rata-rata dari Precision@k pada titik relevan
    return precision_sum / len(relevant_set) if len(relevant_set) > 0 else 0.0

def mean_average_precision(ap_scores: List[float]) -> float:
    """
    Menghitung Mean Average Precision (MAP@k).
    Rata-rata dari Average Precision (AP) dari semua query.
    """
    if not ap_scores:
        return 0.0
    return sum(ap_scores) / len(ap_scores)

# --- Catatan tentang nDCG@k ---
# Implementasi nDCG@k lebih kompleks karena memerlukan skor relevansi bertingkat (graded relevance).
# Karena gold set kita hanya menggunakan biner (relevan/tidak relevan), kita akan fokus pada Precision@k dan MAP@k.
# Jika Anda ingin menerapkan nDCG@k, Anda harus mendefinisikan skor relevansi (misal 3=sangat relevan, 2=cukup, 1=sedikit, 0=tidak).

print("✅ Fungsi Metrik Ranking (Precision@k, AP, MAP@k) siap digunakan.")
````

````python
%%writefile src/vsm_ir.py
# src/vsm_ir.py: Implementasi Vector Space Model (VSM) - Mendukung Pembobotan Ganda (Soal 04 & 05)

import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
from scipy.sparse import csr_matrix
from collections import defaultdict
from typing import Dict, List, Any

class VSM_IR:
    """Kelas untuk mengelola VSM, menghitung TF-IDF, dan Ranking."""

    def __init__(self, processed_docs: Dict[str, List[str]]):
        self.processed_docs = processed_docs
        self.doc_ids = list(processed_docs.keys())
        self.documents_text = [' '.join(tokens) for tokens in processed_docs.values()]

        # Inisialisasi untuk dua model pembobotan
        self.models = {}
        self.models['tfidf_standard'] = self._build_tfidf_matrix(sublinear=False)
        self.models['tfidf_sublinear'] = self._build_tfidf_matrix(sublinear=True)

        # Kita gunakan vectorizer standar sebagai default untuk query transform
        self.vectorizer_standard = self.models['tfidf_standard']['vectorizer']
        self.vocabulary = self.vectorizer_standard.get_feature_names_out()

        print(f"✅ VSM dibangun dengan 2 skema pembobotan.")
        print(f"   Total Vocabulary: {len(self.vocabulary)} term.")

    def _build_tfidf_matrix(self, sublinear: bool):
        """Membangun matriks TF-IDF untuk skema tertentu."""
        # Gunakan TfidfVectorizer yang berbeda untuk setiap skema
        vectorizer = TfidfVectorizer(sublinear_tf=sublinear)
        tfidf_matrix = vectorizer.fit_transform(self.documents_text)

        print(f"   Skema {'Sublinear' if sublinear else 'Standard'} Matrix shape: {tfidf_matrix.shape}")
        return {'vectorizer': vectorizer, 'matrix': tfidf_matrix}

    def transform_query(self, query: str, model_name: str) -> csr_matrix:
        """
        Representasi query sebagai vektor TF-IDF menggunakan vectorizer yang sesuai.
        """
        if model_name not in self.models:
            raise ValueError(f"Model '{model_name}' tidak ditemukan.")

        vectorizer = self.models[model_name]['vectorizer']
        return vectorizer.transform([query])

    def rank_documents(self, query: str, model_name: str, k: int = 5) -> List[Dict[str, Any]]:
        """
        Menghitung Cosine Similarity dan mengambil top-k menggunakan model yang ditentukan.
        """
        if model_name not in self.models:
            raise ValueError(f"Model '{model_name}' tidak ditemukan. Pilihan: {list(self.models.keys())}")

        # 1. Ubah query menjadi vektor TF-IDF
        query_vector = self.transform_query(query, model_name)

        # 2. Ambil matriks dokumen yang sesuai
        doc_tfidf_matrix = self.models[model_name]['matrix']

        # 3. Hitung Cosine Similarity
        similarity_scores = cosine_similarity(query_vector, doc_tfidf_matrix).flatten()

        # Ranking dan pengurutan (sama seperti sebelumnya)
        # Ambil top-k index
        if k > 0 and len(similarity_scores) >= k:
            top_k_indices = np.argpartition(similarity_scores, -k)[-k:]
        else:
            top_k_indices = np.argsort(similarity_scores)[::-1] # Ambil semua jika k terlalu besar

        results = []
        for index in top_k_indices:
            doc_id = self.doc_ids[index]
            score = similarity_scores[index]
            results.append({'doc_id': doc_id, 'score': score, 'index': index, 'model': model_name})

        results.sort(key=lambda x: x['score'], reverse=True)

        return results
````

````python
%%writefile app/chatbot.py
# app/chatbot.py: RAG Mini Chatbot dengan Gradio (FIX: Guardrail Skor)

import gradio as gr
import sys
import os
import glob
from typing import Tuple, List, Dict, Any # Tambahkan List/Dict/Any untuk konsistensi

# Tambahkan path ke folder src agar bisa mengimpor search_engine
sys.path.append(os.path.join(os.getcwd(), 'src'))
import search_engine

# --- Muat Data Mentah (untuk generator) ---
def load_raw_data(src_dir='data/src'):
    raw_docs = {}
    input_dir = os.path.join(os.getcwd(), src_dir)
    for file_path in glob.glob(os.path.join(input_dir, '*.txt')):
        doc_id = os.path.basename(file_path).replace('.txt', '')
        with open(file_path, 'r', encoding='utf-8') as f:
            raw_docs[doc_id] = f.read()
    return raw_docs

# PENTING: Inisialisasi model IR global di awal
raw_documents = load_raw_data()
processed_documents = search_engine.load_processed_data()
search_engine.initialize_ir_systems(processed_documents)


# --- GENERATOR TEMPLATE-BASED (Soal 05, Langkah 3) ---
# Tambahkan placeholder untuk output jika retrieval gagal
FALLBACK_RESPONSE = "Maaf, saya tidak menemukan informasi relevan terkait pertanyaan Anda dalam korpus yang tersedia."

def template_generator(context_text: str, query: str) -> str:
    """
    Generator sederhana: gabungkan konteks dan format jawaban.
    """
    if not context_text:
        return FALLBACK_RESPONSE

    response = (
        f"Berdasarkan dokumen yang relevan, berikut ringkasan informasinya:\n"
        f"----------------------------------------\n"
        f"{context_text}\n"
        f"----------------------------------------\n"
        f"Informasi ini didapat dari pencarian Vector Space Model."
    )
    return response

# --- FUNGSI UTAMA CHATBOT (Interface Gradio) ---
def chatbot_response(query: str) -> Tuple[str, str]:
    """
    Fungsi RAG (Retrieval-Augmented Generation).
    """
    if not query.strip():
        return "Silakan masukkan pertanyaan.", ""

    # 1. RETRIEVAL (Menggunakan VSM dari search_engine)
    k = 3
    retrieval_results: List[Dict[str, Any]] = search_engine.search_query(query, model='vsm', k=k)

    context = ""
    source_info = "Sumber:\n"

    # --- GUARDRAIL PENTING ---
    # Jika skor dokumen teratas (indeks 0) terlalu rendah (misal < 0.1), anggap retrieval gagal.
    # Nilai 0.1 adalah nilai heuristik; Anda bisa menentukannya sendiri.
    RETRIEVAL_THRESHOLD = 0.05

    is_relevant = False
    if retrieval_results and retrieval_results[0]['score'] > RETRIEVAL_THRESHOLD:
        is_relevant = True

    # 2. AUGMENTATION & CONTEXT GATHERING
    if is_relevant:
        context_parts = []
        for i, res in enumerate(retrieval_results):
            doc_id = res['doc_id']
            # Ambil konten mentah dari dokumen yang ditemukan
            doc_content = raw_documents.get(doc_id, "Dokumen tidak ditemukan.")

            # Gabungkan konten ke dalam konteks, ambil 150 karakter pertama sebagai snippet
            context_parts.append(f"Dokumen {doc_id.upper()} (Skor: {res['score']:.4f}): {doc_content[:150]}...")

            source_info += f"- {doc_id.upper()} (Skor: {res['score']:.4f})\n"

        context = "\n\n".join(context_parts)
    else:
        # Jika retrieval gagal, set context ke string kosong dan tampilkan sumber yang diambil (skor rendah)
        context = ""
        if retrieval_results:
             source_info += "Retrieval gagal karena skor tertinggi terlalu rendah:\n"
             for res in retrieval_results:
                 source_info += f"- {res['doc_id'].upper()} (Skor: {res['score']:.4f})\n"
        else:
             source_info += "Tidak ada dokumen yang ditemukan."


    # 3. GENERATION
    generated_answer = template_generator(context, query)

    # Jika context kosong, Generator akan mengembalikan FALLBACK_RESPONSE
    if not is_relevant:
        return FALLBACK_RESPONSE, source_info

    return generated_answer, source_info


# --- INTERFACE GRADIO ---
with gr.Blocks(title="STKI RAG Mini Chatbot") as demo:
    gr.Markdown("# STKI RAG Mini Chatbot (Soal 05)")
    gr.Markdown("Masukkan pertanyaan Anda. Sistem menggunakan VSM untuk retrieval (k=3). **Guardrail aktif untuk skor rendah.**")

    with gr.Row():
        query_input = gr.Textbox(label="Pertanyaan Anda:", placeholder="Contoh: Berapa jam ideal untuk tidur?", lines=2)

    with gr.Row():
        submit_btn = gr.Button("Cari Jawaban (RAG)")

    answer_output = gr.Textbox(label="Jawaban Chatbot (Generator):", lines=10)
    source_output = gr.Textbox(label="Context Retrieval (Sumber VSM):", lines=5)

    submit_btn.click(
        fn=chatbot_response,
        inputs=[query_input],
        outputs=[answer_output, source_output]
    )

if __name__ == "__main__":
    print("\nStarting Gradio App...")
    demo.launch(share=True)
````

````python
%%writefile src/eval.py
# src/eval.py: Modul untuk Evaluasi Model IR (Soal 05)

from typing import List, Set, Dict

def precision_at_k(retrieved_list: List[str], relevant_set: Set[str], k: int) -> float:
    """
    Menghitung Precision@k.
    Proporsi dokumen yang relevan di antara top-k yang diambil.
    """
    if k == 0:
        return 0.0

    retrieved_k = retrieved_list[:k]
    relevant_in_k = len(set(retrieved_k).intersection(relevant_set))

    return relevant_in_k / k

def average_precision(retrieved_list: List[str], relevant_set: Set[str]) -> float:
    """
    Menghitung Average Precision (AP).
    """
    relevant_count = 0
    precision_sum = 0.0

    if not relevant_set:
        return 0.0

    for i, doc_id in enumerate(retrieved_list):
        k = i + 1
        if doc_id in relevant_set:
            relevant_count += 1
            # Hitung Precision@k pada titik ini dan jumlahkan
            precision_sum += precision_at_k(retrieved_list, relevant_set, k)

    return precision_sum / len(relevant_set) if len(relevant_set) > 0 else 0.0

def mean_average_precision(ap_scores: List[float]) -> float:
    """
    Menghitung Mean Average Precision (MAP).
    Rata-rata dari Average Precision (AP) dari semua query.
    """
    if not ap_scores:
        return 0.0
    return sum(ap_scores) / len(ap_scores)

# Anda juga bisa menambahkan fungsi F1-score dan nDCG@k di sini.
def f1_score(precision: float, recall: float) -> float:
    """
    Menghitung F1 Score dari Precision dan Recall.
    Digunakan untuk evaluasi Boolean (Soal 03) dan perbandingan akhir (Soal 05).
    """
    if precision + recall == 0:
        return 0.0
    return 2 * (precision * recall) / (precision + recall)
````
